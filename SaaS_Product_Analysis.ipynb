## **Usage Metrics**

```python
account_engagement_df = (
    account_feature_usage_df
    .groupby("account_id")
    .agg(
        total_events=("usage_date", "count"),
        active_days=("usage_date", "nunique"),
        first_usage_date=("usage_date", "min"),
        last_usage_date=("usage_date", "max")
    )
    .reset_index()
)
```

```python
analysis_date = account_feature_usage_df["usage_date"].max()

account_engagement_df["days_since_last_use"] = (
    analysis_date - account_engagement_df["last_usage_date"]
).dt.days
```

```python
account_engagement_df = account_engagement_df.merge(
    accounts_with_churn_df[["account_id", "is_churned", "churn_date"]],
    on="account_id",
    how="left"
)
```

```python
account_engagement_df["total_events"] = account_engagement_df["total_events"].fillna(0)
account_engagement_df["active_days"] = account_engagement_df["active_days"].fillna(0)
```

```python
account_engagement_df.head(5)
```

# **PHASE - 3 : COHORT & RETENTION ANALYSIS**


```python
df_acc["signup_month"] = (
    df_acc["signup_date"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
account_feature_usage_df["activity_month"] = (
    account_feature_usage_df["usage_date"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
usage_with_cohort_df = account_feature_usage_df.merge(
    df_acc[["account_id", "signup_month"]],
    on="account_id",
    how="left"
)
```

```python
usage_with_cohort_df["signup_month"] = pd.PeriodIndex(
    usage_with_cohort_df["signup_month"], freq="M"
)

usage_with_cohort_df["activity_month"] = pd.PeriodIndex(
    usage_with_cohort_df["activity_month"], freq="M"
)
```

```python
usage_with_cohort_df["month_offset"] = (
    usage_with_cohort_df["activity_month"] -
    usage_with_cohort_df["signup_month"]
).apply(lambda x: x.n)

usage_with_cohort_df = usage_with_cohort_df[
    usage_with_cohort_df["month_offset"] >= 0
]
```

```python
cohort_activity_df = (
    usage_with_cohort_df
    .groupby(["signup_month", "month_offset"])["account_id"]
    .nunique()
    .reset_index(name="active_accounts")
)
```

```python
cohort_activity_df
```

```python
cohort_size_df = (
    df_acc
    .groupby("signup_month")["account_id"]
    .nunique()
    .reset_index(name="cohort_size")
)
```

```python
cohort_size_df["signup_month"] = pd.PeriodIndex(cohort_size_df["signup_month"], freq="M")

cohort_retention_df = cohort_activity_df.merge(
    cohort_size_df,
    on="signup_month",
    how="left"
)
```

```python
cohort_retention_df["retention_rate"] = (
    cohort_retention_df["active_accounts"] /
    cohort_retention_df["cohort_size"]
)
```

```python
support_compare = account_churn_support_df.groupby("is_churned")[[
    "ticket_count",
    "avg_resolution_time"
]].describe()

support_compare
```

```python
churn_driver_summary = account_churn_support_df.groupby("is_churned").agg(
    avg_events=("total_events", "mean"),
    avg_active_days=("active_days", "mean"),
    avg_days_since_last_use=("days_since_last_use", "mean"),
    avg_ticket_count=("ticket_count", "mean"),
    avg_resolution_time=("avg_resolution_time", "mean")
)

churn_driver_summary
```

# **PHASE 5 : REVENUE & BUSINESS IMPACT**

```python
account_revenue_df = (
    df_subs
    .sort_values("mrr_amount", ascending=False)
    .drop_duplicates(subset=["account_id"])
    [["account_id", "mrr_amount", "arr_amount"]]
)
```

```python
revenue_churn_df = accounts_with_churn_df.merge(
    account_plan_df,
    on="account_id",
    how="left"
).merge(
    account_revenue_df,
    on="account_id",
    how="left"
)

revenue_churn_df["plan_tier"] = revenue_churn_df["plan_tier_y"]
```

```python
revenue_churn_df[["mrr_amount", "arr_amount"]] = (
    revenue_churn_df[["mrr_amount", "arr_amount"]].fillna(0)
)
```

```python
total_mrr = revenue_churn_df["mrr_amount"].sum()
active_mrr = revenue_churn_df.loc[
    revenue_churn_df["is_churned"] == 0, "mrr_amount"
].sum()
churned_mrr = revenue_churn_df.loc[
    revenue_churn_df["is_churned"] == 1, "mrr_amount"
].sum()
```

```python
total_mrr = revenue_churn_df["mrr_amount"].sum()
active_mrr = revenue_churn_df.loc[
    revenue_churn_df["is_churned"] == 0, "mrr_amount"
].sum()
churned_mrr = revenue_churn_df.loc[
    revenue_churn_df["is_churned"] == 1, "mrr_amount"
].sum()
```

```python
logo_churn_rate = revenue_churn_df["is_churned"].mean()
revenue_churn_rate = churned_mrr / total_mrr if total_mrr > 0 else 0
```

```python
revenue_churn_by_plan = (
    revenue_churn_df
    .groupby("plan_tier")
    .agg(
        total_mrr=("mrr_amount", "sum"),
        churned_mrr=("mrr_amount", lambda x: x[revenue_churn_df.loc[x.index, "is_churned"] == 1].sum()),
        churned_accounts=("is_churned", "sum"),
        total_accounts=("account_id", "count")
    )
    .reset_index()
)

revenue_churn_by_plan["revenue_churn_rate"] = (
    revenue_churn_by_plan["churned_mrr"] / revenue_churn_by_plan["total_mrr"]
).fillna(0)
```

```python
print(f"Total MRR: {total_mrr:,.0f}")
print(f"Active MRR: {active_mrr:,.0f}")
print(f"Churned MRR: {churned_mrr:,.0f}")
print(f"Logo Churn Rate: {logo_churn_rate:.2%}")
print(f"Revenue Churn Rate: {revenue_churn_rate:.2%}")
revenue_churn_by_plan
```

```python
retention_matrix = cohort_retention_df.pivot_table(
    index="signup_month",
    columns="month_offset",
    values="retention_rate"
)
```
